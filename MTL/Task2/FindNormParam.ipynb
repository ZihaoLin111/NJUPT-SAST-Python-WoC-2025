```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
```

```python
dataset = datasets.CIFAR10(
    root='./data', 
    train=True,
    download=True,
    transform=transforms.ToTensor()
)
```

```python
loader = DataLoader(dataset, batch_size=1024, shuffle=False, num_workers=4)
```

```python
sum_rgb = torch.tensor([0.0, 0.0, 0.0])
sum_squared_rgb = torch.tensor([0.0, 0.0, 0.0])
num_pixels = 0
```

```python
for images, _ in loader:
    batch_size = images.size(0)
    height = images.size(2)
    width = images.size(3)
    
    num_pixels += batch_size * height * width
    
    sum_rgb += torch.sum(images, dim=[0, 2, 3])
    sum_squared_rgb += torch.sum(images ** 2, dim=[0, 2, 3])
```

```python
mean_rgb = sum_rgb / num_pixels
std_rgb = torch.sqrt((sum_squared_rgb / num_pixels) - (mean_rgb ** 2))
```

数学公式
$$\sigma = \sqrt{E[X^2] - (E[X])^2}$$

```python
print("Mean RGB:", mean_rgb)
print("Std RGB:", std_rgb) 
# Mean RGB: tensor([0.4914, 0.4822, 0.4465])
# Std RGB: tensor([0.2470, 0.2435, 0.2616])
```
